Este código utiliza el algoritmo de poda alfa-beta en el juego del tic-tac-toe. Puedes probarlo tal como está.



https://www.youtube.com/watch?v=I0y-TGehf-4

```python
import math

# Contador global para medir el rendimiento de la búsqueda
nodos_visitados = 0

def imprimir_tablero(tablero):
    for fila in tablero:
        print(" ".join(fila))
    print()

def es_tablero_lleno(tablero):
    for fila in tablero:
        if " " in fila:
            return False
    return True

def evaluar_estado(tablero):
    # Evaluación simple para el tic-tac-toe
    ganador = None
    # Filas
    for fila in tablero:
        if fila.count("X") == 3:
            ganador = "X"
        elif fila.count("O") == 3:
            ganador = "O"
    # Columnas
    for col in range(3):
        if tablero[0][col] == tablero[1][col] == tablero[2][col] and tablero[0][col] != " ":
            ganador = tablero[0][col]
    # Diagonales
    if tablero[0][0] == tablero[1][1] == tablero[2][2] and tablero[0][0] != " ":
        ganador = tablero[0][0]
    if tablero[0][2] == tablero[1][1] == tablero[2][0] and tablero[0][2] != " ":
        ganador = tablero[0][2]

    if ganador == "X":
        return 1
    elif ganador == "O":
        return -1
    elif es_tablero_lleno(tablero):
        return 0
    else:
        return None

def obtener_movimientos(tablero):
    # Casillas vacías del tablero
    return [(i, j) for i in range(3) for j in range(3) if tablero[i][j] == " "]

def alphabeta(tablero, profundidad, alfa, beta, es_maximizando, usar_poda=True):
    global nodos_visitados
    nodos_visitados += 1
    estado_actual = evaluar_estado(tablero)

    # nuestra heuristica ganar pronto vale más, perder tarde vale menos
    if estado_actual is not None:
        if estado_actual == 1:
            return 10 - profundidad
        elif estado_actual == -1:
            return profundidad - 10
        return 0

    if es_maximizando:
        mejor_valor = -math.inf
        for i, j in obtener_movimientos(tablero):
            tablero[i][j] = "X"
            valor = alphabeta(tablero, profundidad + 1, alfa, beta, False,
                              usar_poda)
            tablero[i][j] = " "
            mejor_valor = max(mejor_valor, valor)
            alfa = max(alfa, mejor_valor)
            if usar_poda and beta <= alfa:
                break
        return mejor_valor
    else:
        mejor_valor = math.inf
        for i, j in obtener_movimientos(tablero):
            tablero[i][j] = "O"
            valor = alphabeta(tablero, profundidad + 1, alfa, beta, True,
                              usar_poda)
            tablero[i][j] = " "
            mejor_valor = min(mejor_valor, valor)
            beta = min(beta, mejor_valor)
            if usar_poda and beta <= alfa:
                break
        return mejor_valor

def mejor_movimiento(tablero, jugador="O", usar_poda=True):
    # X maximiza y O minimiza (corrige el error del código que mencione antes)
    es_x = jugador == "X"
    mejor_valor = -math.inf if es_x else math.inf
    mejor_mov = None
    alfa = -math.inf
    beta = math.inf

    for i, j in obtener_movimientos(tablero):
        tablero[i][j] = jugador
        valor = alphabeta(tablero, 1, alfa, beta, not es_x, usar_poda)
        tablero[i][j] = " "

        if (es_x and valor > mejor_valor) or (not es_x and valor < mejor_valor):
            mejor_valor = valor
            mejor_mov = (i, j)
        if es_x:
            alfa = max(alfa, mejor_valor)
        else:
            beta = min(beta, mejor_valor)

    return mejor_mov

#nodos visitados con y sin poda desde el tablero vacío
for usar_poda in [False, True]:
    nodos_visitados = 0
    mov = mejor_movimiento([[" "] * 3 for _ in range(3)], "X", usar_poda)
    print(f"Poda={usar_poda}: movimiento {mov}, nodos visitados = {nodos_visitados}")
print()

# Selección del modo de juego
modo = input("Modo de juego (1 = humano vs IA, 2 = humano vs humano): ").strip()

tablero_inicial = [
    [" ", " ", " "],
    [" ", " ", " "],
    [" ", " ", " "]
]

turno = "X"

while evaluar_estado(tablero_inicial) is None:
    imprimir_tablero(tablero_inicial)

    # el turno humano siempre X, y también O en modo multijugador local
    if turno == "X" or modo == "2":
        try:
            fila, col = map(int, input(f"Turno de {turno}. Ingrese fila y \
            columna (separadas por espacio): ").split())
        except ValueError:
            print("Entrada inválida. Intenta de nuevo.")
            continue
        if not (0 <= fila < 3 and 0 <= col < 3):
            print("Fuera del tablero. Intenta de nuevo.")
            continue
        if tablero_inicial[fila][col] == " ":
            tablero_inicial[fila][col] = turno
        else:
            print("Casilla ocupada. Intenta de nuevo.")
            continue
    else:
        print("Turno de O (IA)")
        fila, col = mejor_movimiento(tablero_inicial, "O")
        tablero_inicial[fila][col] = "O"

    turno = "O" if turno == "X" else "X"

imprimir_tablero(tablero_inicial)
resultado = evaluar_estado(tablero_inicial)

if resultado == 1:
    print("¡X gana!")
elif resultado == -1:
    print("¡O gana!")
else:
    print("¡Empate!")
```
